*datetime.date*
---
연,월,일 날짜를 표현할 때 사용하는 함수이고, 주로 사용되는 경우:

1. 현재 시간과 날짜를 구할 때  
2. 날짜의 차이를 구할 때  
3. 특정 날짜의 요일을 구할 때  



```
#기초 예제
import datetime

#현재 시간을 구하기
print(datetime.datetime.now())

#날짜의 차이를 구하기
day1 = datetime.date(2004, 4, 18)
day2 = datetime.date(2025, 11, 15)
diff= day2 - day1
print(diff.days)

#요일 구하기
day= datetime.date(2025, 11, 5)
print(day.weekday())
```
1.현재시간 : 2025-11-05 08:35:47.875274  
2.날짜의 차이 : 7881  
3.요일 : 2  


*time*
---
time 모듈에는 함수가 매우 많음  
1. time.time  
2. time.loclatime  
3. time.asctime  
4. time.ctime  
5. time.strftime  
6. time.sleep


```
#기초 예제
import time

#1970년 1월 1일 0시 0분 0초를 기준으로 지난 시간을 초 단위로 리턴해줌
print(time.time())

#time.time()이 리턴한 실숫값을 사용해서 연,월,일,시,분,초, ...의 형태로 바꾸어줌
print(time.localtime(time.time()))

#time.localtime이 리턴된 튜플 형태의 값을 보기 좋게 리턴
print(time.asctime(time.localtime(time.time())))

#time.asctime()을 간단하게 표시(asctime과 다른점은 현재 시간만 리턴)
print(time.ctime())

#strftime 함수는 시간과 관계된 것을 세밀하게 표현하는 여러 포맷 코드를 제공함
#현재 시간을 기준으로 할 경우 time.strftime('포맷_코드')로 생략가능
#요일
print(time.strftime('%A'))

포맷 코드
(1) %W : 1년 중 누적 주 []
(2) %w : 숫자로 된 요일 []
(3) %j : 1년 중 누적 날짜 []
(4) %d : 일 [01-31]
(5) %Y : 연도 [2025]
(6) %B : 달 [November]
(7) %c : 날짜와 시간 [Thu Nov  6 16:49:00 2025]
```
1 : 1762415657.311418  
2 : time.struct_time(tm_year=2025, tm_mon=11, tm_mday=6, tm_hour=16, tm_min=56, tm_sec=54, tm_wday=3, tm_yday=310, tm_isdst=0)  
3 : Thu Nov  6 16:57:15 2025  
4 : Thu Nov  6 16:57:15 2025

*math*
---

최대공약수와 최소공배수를 구할 수 있다


```
import math
#최대 공약수 구하기(gcd : greatest common divisor)
print(math.gcd(60,100,80))

#최소 공배수 구하기(lcm : least common multiple)
print(math.lcm(15,25))
```
1.최대 공약수 : 20  
2.최소 공약수 : 75

*random*
--

난수를 발생시키는 모듈



```
import random
#0.0에서 1.0 사이의 실수 중 난수 값을 리턴
print(random.random())

#1에서 10사이의 정수 중 난수 값을 리턴
print(random.randint(1, 10))  
```
1 : 0.15362916188184816  
2 : 3


````python
#자동으로 로또 숫자 선택하기
import random
number_list = list(range(1,46))
pop_list= []
while len(pop_list) < 6:
  number= random.randint(0, len(number_list)-1)
  popped_number= number_list.pop(number)
  pop_list.append(popped_number)
print(pop_list)


````

````python
#random.sample()을 사용하면 더 간단하게 사용가능
import random
number_list = range(1, 46)
result_list = random.sample(number_list, 6)

print(result_list)
````
