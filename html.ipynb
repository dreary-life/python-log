tag
--
html의 문법은 한 가지, <font color="red">태그(tag)</font>이다.  
tag는 <태그명> 콘텐츠 기입 </태그명>의 형태이다.  
태그를 통해 시작과 끝을 표시하는데, <태그명 /> 또는 <태그명>은 단일 태그로 시작만 표시할 수 있다.

속성
---

속성은 태그의 부가적인 기능을 정의하며 시작 태그의 내부에 정의한다.  
속성의 개수에는 특별한 제한이 없다.

속성이 추가된 태그의 예)  
<태그명 속성명="속성값"> 내용 </태그명>  
<태그명 속성명="속상값" />

#html의 기본구조

```
<!DOCTYPE html>
<html>
  <head>  
    <meta charset="utf-8">   
    <title> 문서의 제목을 쓰는 곳  </title>
  </head>
  <body>
    브라우저 화면에 나올 내용을 작성하는 곳
  </body>

</html>
```
```
<!DOCTYPE html> : 문서 유형을 지정하는 단일 태그
<html> </html> : 문서의 시작과 끝을 나타내는 태그
<head> </head> : 화면에는 보이지 않지만 웹 브라우저가 알아야할 할 정보들 기입
<meta charset="utf-8"> : 문자 인코딩 및 문서 키워드 등에 대한 정보를 기입하는 단일 태그
<title> </title> : 문서의 제목, 콘텐츠는 브라우저 탭에 표시된다.
<body> </body> : 실제 브라우저 화면에 표시될 내용을 입력하는 태그
